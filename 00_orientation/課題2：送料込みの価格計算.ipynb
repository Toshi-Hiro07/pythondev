```python
fee = int(input("税抜価格を入力してください"))
print("税抜価格を入力してください："+str(fee))
if fee>=2000:
    print("送料は無料です")
    fee=fee*1.1
    fee=int(fee)
    print("送料込みの価格は"+str(fee)+"です。")
if fee<2000:
    fee=fee*1.1+350
    fee=int(fee)
    print("送料として350円かかります")
    print("送料込みの価格は"+str(fee)+"です。")
```
