```python
letters = "abcdefghijklmnopqrstuvwxyz"
alphabet_dict = {ch: 0 for ch in letters}

words = []

while True:
    word = input("英単語を入力してください（終了はEnter）：")
    if word == "":
        break
    words.append(word)  
    for ch in word.lower():   
        if ch in alphabet_dict:
            alphabet_dict[ch] += 1
print("\n入力した英単語:")
print(sorted(words))

for k, v in sorted(alphabet_dict.items()):
    if v > 0: 
        print(k, "が", v, "個ありました")


```
